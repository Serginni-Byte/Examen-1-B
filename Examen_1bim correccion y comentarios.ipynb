```python
import pandas as pd
df_clientes = pd.read_csv("clientes.csv",sep = ',', header = 0, encoding =  'utf-8' )

display(df_clientes)

df_productos = pd.read_csv("productos_.csv",sep = ',', header = 0, encoding =  'utf-8' )

display(df_productos)

df_ventas = pd.read_csv("ventas.csv",sep = ',', header = 0, encoding =  'utf-8',parse_dates=['fecha'] )

display(df_ventas)
```

1. **Líneas, tickets y unidades.** ¿Cuántas líneas de venta, cuántos tickets distintos y cuántas unidades vendidas hay en el archivo? Explique qué representa cada conteo.


```python
linea_ventas = len(df_ventas)

tickets_distintos = df_ventas['id_ticket'].nunique()

unidades_vendidas = df_ventas['cantidad'].sum()

tabla1 = pd.DataFrame({
    'Preguntas': [
        'Total lineas de venta', 
        'Total tickets distintos', 
        'Total unidades vendidas'
    ],
    'Cantidad': [
        linea_ventas, 
        tickets_distintos, 
        unidades_vendidas
    ]
})

tabla = tabla1.set_index('Preguntas')

display(tabla)
```

2. **Tickets por sucursal.** ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?


```python
tickets_sucursal = df_ventas.groupby('sucursal')['id_ticket'].agg(['nunique']).rename(columns={'nunique':'Tickets registrados'})

display(tickets_sucursal)
```

3. **Métodos de pago.** ¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?


```python
metodo_pago = df_ventas.groupby('metodo_pago')['id_ticket'].agg(['nunique']).rename(columns={'nunique':'Tickets registrados'})

totalMetodos = metodo_pago['Tickets registrados'].sum()

display(metodo_pago)

print("Hay",tickets_distintos-totalMetodos,"pagos sin metodo registrado")
```

4. **Precios de lista por categoría.** ¿Cuál es el precio de lista mínimo, máximo y promedio de los productos de cada categoría? Use el catálogo de productos.

```python
lista_categoria_precio = df_productos.groupby('categoria')['precio_lista'].agg(['min','max','mean']).round(2).rename(columns={'min':'Precio mínimo [$]','max':'Precio máximo [$]','mean':'Promedio precio [$]'})

display(lista_categoria_precio)
```

CORRECCION 7. **Tamaño de las compras.** ¿Cuántos tickets corresponden a compras de **hasta 7 unidades**, **de 8 a 14 unidades** y **de 15 o más unidades**? Sume primero las cantidades de todas las líneas de cada ticket.

CORRECCION 8. **Ventas por mes.** ¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?

```python
#Mi error fue más lógico y también de sintaxis, desconocía como obtener simplemente el mes
#Creamos una nueva columna temporal llamada mes, que es igual a la columna de la fecha pero solo con el mes 
#para ello usamos .dt.to_period('') dentro de las comillas colocamos la parte de la fecha que deseamos

df_ventas['mes'] = df_ventas['fecha'].dt.to_period('M')

#Sabía que tenia que usar un groupby pues todo se clasifica por el mes, sin embargo 
#estaba algo confundido respecto a como colocar las otras columnas, podemos para ello
#dentro del .agg crear dos columnas y dentro de ellas dictaer que funciones se van a aplicar
#y todo eso se clasificará respecto al mes con el groupby anterior.
#Usamos reset_index() de forma opcional, nos permite ver mejor que existen 12 meses en la tabla

ventas_mes = df_ventas.groupby('mes').agg(Tickets_Registrados=('id_ticket','nunique'),Importes_Registrado=('importe','sum')).reset_index()

display(ventas_mes)

#Ordenamos las ventas, podiamos nosotros ver facilmente cual era el importe mayor, pero en la vida real
#podemos tener más fechas y se volvería dificil, para ello es mejor sortearlo como dice abajo
#lo hacemos "by" la columna de los importes, de forma descendente, o sea, primero el mayor luego el menor
#asi podemos usar un .iloc(0) para obtener el mayor
ventas_mes_sort = ventas_mes.sort_values(by='Importes_Registrado',ascending=False)

mayor_importe = ventas_mes_sort.iloc[0]

print("El mes con mayor importe es:",mayor_importe['mes'], ", con un total de:",mayor_importe['Importes_Registrado'],)

```

CORRECCION 9. **Producto más vendido por sucursal.** ¿Qué producto acumuló más unidades vendidas en cada sucursal? Indique su código, nombre y cantidad vendida. Si hay un empate, incluya todos los productos empatados.

```python

#Mi error nuevamente estaba en no saber utilizar el merge, desconocer su sintaxis.
#El objetivo del merge es usar el nombre del producto y no solo su id, para ello las mezclamos en
#la columna que tienen en comun, que es id_producto, usamos "left" para que df_ventas sea el protagonista
otro_merge = pd.merge(df_ventas,df_productos,on='id_producto',how="left")

#Aqui mi error estuvo en no saber que un groupby podía contener multiples columnas, es algo que pude 
#incluso haber aplicado en otras partes, es bueno saberlo.
#Luego de aplicar un multiple groupby, sumamos las cantidades y reseteamos el indice
#este reseteo es importante para poder aplicar la comparacion logica de orden_sucursal
producto_por_sucursal = otro_merge.groupby(['sucursal','id_producto','producto'])['cantidad'].sum().reset_index()

#Aqui debemos ordenar las cantidades de producto_por_sucursal que sean igual a las cantidades con un valor máximo por sucursal
#es decir, el groupby nos debe devolver solo tres filas, pero queremos que sean las tres filas con la maxima cantidad vendida por sucursal
#como en el SUR hay dos valores maximos se implementan esos dos tambien ya que cumplen la igualdad.

#Aqui yo desconocía la funcion .transform('max'), pero tampoco tenia idea de que podría
#aplicar una logica de hallar un valor que sea igual al valor maximo y solo tomar eso.
#Asi orden_sucursal no extrae las cientos de filas, solo cuya cantidad por sucursal sea igual a la cantidad maxima por sucursal
orden_sucursal = producto_por_sucursal[producto_por_sucursal['cantidad'] == producto_por_sucursal.groupby('sucursal')['cantidad'].transform('max')]

#Aqui ordenamos con un sort por sucursal, luego reseteamos el indice con un reset_index o si no guardará los indices originales, que se vería raro
#y también renombramos la columna cantidad por algo más intuitivo
orden_maximo = orden_sucursal.sort_values(by='sucursal',ascending=True).reset_index(drop=True).rename(columns={'cantidad': 'Unidades Vendidas'})


display(orden_maximo)
```

CORRECCION 10. **Compras por tipo de cliente.** Para cada tipo de cliente, ¿cuántos clientes distintos compraron, cuántos tickets generaron y cuál fue el importe total de sus compras?

```python
#Mi error era no conocer pd.merge


#Una vez conoci pd.mergey más a fondo la funcion .agg en ejercicios anteriores, lo unico que tuve que hace
#es combinar dos tablas en el id del cliente donde la tabla izquierda sea la principal.
#como en ambos csv la columna del id del cliente es distinta entonces uso left_on y right_on

df_ventas2 = pd.merge(df_ventas,df_clientes,left_on="id_cliente",right_on="id",how="left")

#uso la funcion .agg luego de agrupar todo por el tipo de cliente, uso nunique para los tickets generados y los id_clientes ya que no debe repetirse lo unico
#preguntas mas atras yo ya habia creado df_ventas['importe'] asi que solamente he tenido que llamar a esa columna sin problema

compra_per_client = df_ventas2.groupby('tipo_cliente').agg(Clientes_Distintos=('id_cliente','nunique'),Tickets_Generados=('id_ticket','nunique'),Importe_Total=('importe','sum')).rename_axis(index="Tipo Cliente")

display(compra_per_client)
```
